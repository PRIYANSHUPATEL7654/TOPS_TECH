```python
import pandas as pd 
import numpy as np 
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
df = pd.read_csv("student_data.csv")
df.head()
```

```python
print(df.shape)
df.info()
df.describe()
```

```python
X = df[["Age",'Marks','Study_Hours']]
Y = df["Pass"]
print(X.head())
print(Y.head())
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(
    X,Y,test_size=0.2,random_state=42
)

print(X_train.shape)
print(X_test.shape)
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
model = Sequential() # Generates a sequential pipeline of input layer then hidden layers and then output layer.

# Here input - 8 neurons - hidden layer1 (4 neurons)- hidden layer2 - output1
# If there is a larger dataset then we can start from 64,32,16,8,4,2,1 neurons sequentially and accordingly will be the hidden layers and at last will be the output.
model.add(Dense(8,
                activation="relu",
                input_shape=(3,))) # 3 columns as input

model.add(Dense(4,
                activation='relu'))

model.add(Dense(1,
                activation='sigmoid'))
```

```python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
history = model.fit(
    X_train,
    Y_train,
    epochs=50,
    batch_size=8,
    validation_split=0.2   # 20% of training data used for validation
)
```

```python
loss, accuracy = model.evaluate(X_test, Y_test)

print("Loss :", loss)

print("Accuracy :", accuracy)
```

```python
prediction = model.predict(X_test)

prediction
```

```python
new_student = [[22,85,7]]

new_student = scaler.transform(new_student)

result = model.predict(new_student)

if result > 0.5:
    print("Pass")
else:
    print("Fail")
```
