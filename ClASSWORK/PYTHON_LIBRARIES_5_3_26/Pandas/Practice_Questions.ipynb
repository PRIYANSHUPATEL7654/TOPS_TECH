```python
# 10) Create a dataset with duplicate rows.
data1 = pd.DataFrame({
   "Name" : ["Priya","Krisha","Krisha"],
   "Age" : [20,22,22],
   "City" : ["Ahmedabad","Rajkot","Rajkot"],
   "Salary" : [10000,30000,30000]
})
# Tasks:
# • Identify duplicates
data1.duplicated()
# • Remove duplicate rows
data1.drop_duplicates() 
data1
```

```python
# 11) Create a Sales column:
# [100, 120, 110, 115, 500]
Sales = pd.Series([100, 120, 115, 500])

# Tasks:
# • Detect outliers using IQR method
Q1 = Sales.quantile(0.25)
Q3 = Sales.quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
outliers = Sales[(Sales < lower) | (Sales > upper)]
outliers
```

```python
# 12) Using the same sales data:

# • Calculate Z-score for each value  
mean = Sales.mean()
std = Sales.std()

z_scores = (Sales - mean) / std

z_scores
# • Identify values where Z-score > 2 
z_scores[z_scores > 2]
```

```python
# Final Integrated Question (Important)

# 15) Mini Data Cleaning & Analysis Task

# Create a dataset:
# Customer, Age, City, Price, Quantity
# Include:
# • Missing values in Age and City  
# • Duplicate rows  
data3 = pd.DataFrame({
    "Customer": ["Amit", "Riya", "Karan", "Neha", "Riya"],
    "Age": [25, np.nan, 30, 22, np.nan],
    "City": ["Ahmedabad", "Surat", None, "Rajkot", "Surat"],
    "Price": [1000, 2000, 1500, 1800, 2000],
    "Quantity": [2, 3, 1, 4, 3]
})

# Tasks:
# 1 Clean missing values  
#    • Replace Age with mean
data3["Age"] = data3["Age"].fillna(data3["Age"].mean())
#    • Replace City with "Unknown"
data3["City"] = data3["City"].fillna("Unknown")

# 2 Remove duplicate rows  
non_dup_data = data3.drop_duplicates()
non_dup_data
# 3 Create new column:  
#    Revenue = Price * Quantity
data3["Revenue"] = data3["Price"] * data3["Quantity"]
data3
# 4 Filter rows where Revenue > 5000  
data3[data3["Revenue"] > 5000]

# 5 Sort the dataset by Revenue in descending order 
sorted = data3.sort_values(by="Revenue",ascending=False)

# 6 Reset the index  
sorted.reset_index()

# 7 Identify any outliers in Revenue using IQR method
Q1 = data3["Revenue"].quantile(0.25)
Q3 = data3["Revenue"].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = data3["Revenue"][(data3["Revenue"] < lower) | (data3["Revenue"] > upper)] # here we cannot use normal "or" because it gives error as "or" only checks true_false for 1 time but here in series we need to check it again and again so we use "|" and not "or". 
outliers
```
