```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
```

```python
df = pd.read_csv("student_data_knn_100_records.csv")
print(df.head())
```

```python
X = df[["Height","Salary"]]
Y = df["Category"]
print(X.head())
```

```python
scaler = MinMaxScaler()

X_scaled = scaler.fit_transform(X) # transforms in the range of 0 to 1

print(X_scaled[:5])
```

```python
model = KNeighborsClassifier(n_neighbors=3) # we only take odd numbers to get output because even numbers would give a tie

model.fit(X_scaled,Y)

print("Model trained successfully")
```

```python
new_person = pd.DataFrame({
    "Height" : [175],
    "Salary" : [70000]
})

new_person_scaled = scaler.transform(new_person)

prediction = model.predict(new_person_scaled)

print("Prediction : ",prediction[0])
```
