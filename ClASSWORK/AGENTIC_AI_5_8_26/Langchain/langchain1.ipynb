```python
# simple 

# user question - sent to langchain - sent to llm - gives answer - output given to user
# what is langchain framework - it is a framework of agentic ai, it only performs single task at a time 

# Needs :
# gemini api key 
# The gemini api key is sensitive and if it gets exposed then google deletes it so make sure to store it in .env file.
#Gemini 3.5 Flash
# env 
```

```python
from dotenv import load_dotenv # to import the gemini api key
import os

# os - os 
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.messages import HumanMessage
```

```python
load_dotenv()

# api key access 

# question - langchain - llm 
# model version : eg, gemini 2.5 flash
# temperature : Ranges from 0 to 0.15, as the temperature increases the model becomes more descriptive.
# 0 = cut 2 cut answer 
# average descriptive generation : 0.7 
# more than average descriptive generation : 0.10 
```

```python
llm=ChatGoogleGenerativeAI(

    model="gemini-3.5-flash",
    temperature=0.7
)
```

```python
question= input("Enter your question : ")
```

```python
# invoke works like api 
response =  llm.invoke([HumanMessage(content=question)]) # takes the question from user to llm 
# question - langchain - llm - answer =print 
# invoke method passes human message to llm
```

```python
print(response.content)
```
