```python
# 1. Write a short Python script that generates a random dataset of 10 users, each with two features:
# Number of daily app opens
# Average session duration (minutes)
# Print the dataset as a list of dictionaries.

import random

dataset = []

for i in range(1, 11):
    user = {
        "User_ID": f"User_{i}",
        "Daily_App_Opens": random.randint(1, 30),
        "Average_Session_Duration": round(random.uniform(2, 60), 2)
    }
    dataset.append(user)

print("Random Dataset:\n")

for user in dataset:
    print(user)

# random.randint(1,30) generates random daily app opens.
# random.uniform(2,60) generates session duration in minutes.
# Each user is stored as a dictionary.
# All dictionaries are stored in a list.
```
