```python
from dotenv import load_dotenv
import os 

from langgraph.graph import StateGraph, START, END
from typing import TypedDict # data gets in python dictionary with this 
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
load_dotenv()
```

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-3.5.flash",
    temperature=0
    )
```

```python
class State(TypedDict): # this is state class where we store all our variables such as name and this class will be accessed by nodes in this file
    question:str
    answer:str 

# inheritance : child class can access the function of parent class
```

```python
def llm_node(state: State):

    question = state["question"]

    prompt = f"""
    Answer the following question in exactly 100 words.
    Do not write more or fewer than 100 words.
    Question:{question}
    """

    response = llm.invoke(prompt)

    return{
        "answer": response.content
    }
```

```python
graph = StateGraph(State)
```
