```python
# Load the Iris dataset using sklearn, apply PCA to reduce it to 2 principal components, and print the explained variance ratio for both components.

from sklearn.datasets import load_iris
from sklearn.decomposition import PCA

# Load Iris dataset
iris = load_iris()
X = iris.data

# Apply PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

# Print explained variance ratio
print("Explained Variance Ratio:")
print(pca.explained_variance_ratio_)
```
