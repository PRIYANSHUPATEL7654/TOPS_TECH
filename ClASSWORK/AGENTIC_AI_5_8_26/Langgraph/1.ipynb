```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict # data gets in python dictionary with this 
```

```python
class State(TypedDict): # this is state class where we store all our variables such as name and this class will be accessed by nodes in this file
    name:str 

# inheritance : child class can access the function of parent class
```

```python
def mydata(state:State): # here in langgraph functions are called nodes like in langchain  every function was called node
    return{
        "name" : "testing node"
    }
```

```python
graph = StateGraph(State)
```

```python
graph.add_node("hello",mydata)
```

```python
graph.add_edge(START,"hello")
graph.add_edge("hello",END)
```

```python
app = graph.compile()
```

```python
result = app.invoke(
    {
        "name":""
    }
)
print(result["name"])
```
