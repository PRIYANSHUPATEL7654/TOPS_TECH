```python
# Change the linkage method in your clustering code from 'single' to 'complete' and 'average', and observe how the dendrogram structure changes for your songs dataset.<br><br><em><strong>Hint:</strong> Use the linkage parameter in AgglomerativeClustering or scipy's linkage function.</em>

import numpy as np
import matplotlib.pyplot as plt

from scipy.cluster.hierarchy import linkage,dendrogram

songs=np.array([

[120,0.80],

[118,0.75],

[90,0.30],

[95,0.35],

[130,0.90],

[135,0.88],

[80,0.20],

[85,0.25]

])

methods=['single','complete','average']

plt.figure(figsize=(18,5))

for i,method in enumerate(methods):

    plt.subplot(1,3,i+1)

    Z=linkage(
        songs,
        method=method
    )

    dendrogram(
        Z,
        labels=['S1','S2','S3','S4','S5','S6','S7','S8']
    )

    plt.title(method.capitalize()+" Linkage")

plt.tight_layout()

plt.show()
```
