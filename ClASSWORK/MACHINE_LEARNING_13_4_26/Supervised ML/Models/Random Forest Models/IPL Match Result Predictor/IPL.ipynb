```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score,classification_report
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv("IPL.csv")
# Literal value means the value that can never be changed
print(df.head())
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum() # it searches for any duplicate values
```

```python
le = LabelEncoder()

df["season"] = df["season"].astype(str).str.split("/").str[0]
df["season"] = df["season"].astype(int)

df["event_match_no"] = df["event_match_no"].replace("Unknown", np.nan)
df["event_match_no"] = pd.to_numeric(df["event_match_no"])
df["event_match_no"] = df["event_match_no"].fillna(df["event_match_no"].median())

for col in df.columns:
    unique_types = df[col].dropna().apply(type).unique()

    if len(unique_types) > 1:
        print(f"Column: {col}")
        print("Types:", unique_types)
        print()

for col in df.select_dtypes(include='object').columns:
    df[col] = LabelEncoder().fit_transform(df[col].astype(str))
```

```python
X = df.drop("match_won_by",axis=1)
Y = df["match_won_by"]
print(X.head())
print(Y.head())
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(
    X,Y,test_size=0.2,random_state=42
) # this random state is for the data so the training and testing data does not get mixed
```

```python
model = RandomForestClassifier(
    n_estimators=150,
    max_depth=20,
    random_state=42
) # this random state is for the tree so the training and testing data does not get mixed

model.fit(X_train,Y_train)
```

```python
Y_pred = model.predict(X_test)
accuracy_score(Y_test, Y_pred)
```
