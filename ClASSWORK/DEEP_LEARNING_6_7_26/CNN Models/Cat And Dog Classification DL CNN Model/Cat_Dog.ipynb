```python
import tensorflow as tf 
import matplotlib.pyplot as plt
import numpy as np 
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D
from tensorflow.keras.layers import Flatten, Dense
```

```python
train_data = ImageDataGenerator(rescale=1./255)
# ImageDataGenerator (class of tensorflow) is used to load the image
# Image pixels range from 0 to 255 (0 means black and 1 means white)
# Eg, 120 189 200 245
#     230 243 123 230
# This all numbers will be divided by 255 to convert it between 0 to 1
# So the converted pixels might be like below
# Eg, 0.2 0.3 0.5 0.7
#     0.3 0.4 0.6 0.8
test_data = ImageDataGenerator(rescale=1./255)

train = train_data.flow_from_directory(
    "C:/MY DATA/GITHUB DATA/TOPS_TECH/ClASSWORK/DEEP_LEARNING_6_7_26/CNN Models/Cat And Dog Classification DL CNN Model/training",
    target_size=(128,128), # it means image of any size (eg, 500*500) will be converted to 128*128
    batch_size=32, #if the images are 8000 and we have set the batch size then 8000/32 will be the total number of data taken in each epoch and epoch will be set manually to increase accuracy
    class_mode="binary"
)

test = test_data.flow_from_directory(
    "C:/MY DATA/GITHUB DATA/TOPS_TECH/ClASSWORK/DEEP_LEARNING_6_7_26/CNN Models/Cat And Dog Classification DL CNN Model/testing",
    target_size=(128,128), # it means image of any size (eg, 500*500) will be converted to 128*128
    batch_size=32,
    class_mode="binary"
)
```

```python
model = Sequential()
# It will create a sequential pipeline
model.add(Conv2D(
    filters=32, # It will find 32 filters and each image will be checked for all this filters if they are present in it or not. 
    kernel_size=(3,3), # It will create a different combinations of window of 3*3 and max or average pooling is done from it.
    activation="relu",
    input_shape=(128,128,3) # (width, height, color pallete)
))

model.add(MaxPooling2D(pool_size=(2,2)))
# After pooling the matrix we get is called feature map.

model.add(Conv2D(
    filters=64, # By increasing filters the recognisation will increase.
    kernel_size=(3,3),
    activation="relu",
))

model.add(MaxPooling2D(pool_size=(2,2)))

model.add(Flatten()) # It will convert 2D matrix into 1D.

model.add(Dense(128, activation='relu'))

model.add(Dense(1, activation='sigmoid'))
```

```python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
history = model.fit(
    train,
    validation_data=test,
    epochs=10
)
```

```python
loss,accuracy = model.evaluate(test)
print("Accuracy : ",accuracy)
```
