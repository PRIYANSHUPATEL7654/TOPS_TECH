```python
# Create a scatterplot of Annual Income vs Spending Score using matplotlib or seaborn to visualize how customers are distributed.<br><br><em><strong>Hint:</strong> Use seaborn.scatterplot() or plt.scatter().</em>

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("Mall_Customers.csv")

# Scatter Plot
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=df,
    x='Annual Income (k$)',
    y='Spending Score (1-100)',
    color='blue'
)

plt.title("Annual Income vs Spending Score")
plt.xlabel("Annual Income (k$)")
plt.ylabel("Spending Score (1-100)")
plt.grid(True)

plt.show()
```
