```python
# Use pandas to check for missing values in the Age, Annual Income, and Spending Score columns, and print the total number of missing values for each.

import pandas as pd

# Load dataset
df = pd.read_csv("Mall_Customers.csv")

# Check missing values
missing_values = df[['Age', 'Annual Income (k$)', 'Spending Score (1-100)']].isnull().sum()

print("Missing Values in Selected Columns:\n")
print(missing_values)
```
