```python
import numpy as np
```

```python
X1=2
X2=3
X3=1

W1=0.5
W2=0.4
W3=0.8

Bias = 0.2
```

```python
Z = (X1 * W1) + (X2*W2) + (X3*W3) + Bias
print(Z) # weighted sum
```

```python
def sigmoid(x):
    return 1/(1+np.exp(-x))
```

```python
output = sigmoid(Z)
print(output)
```

```python
if output >= 0.5:
    print("Prediction = 1")
else:
    print("Prediction = 0")
```
