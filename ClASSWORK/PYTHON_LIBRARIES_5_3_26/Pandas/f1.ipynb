```python
# 27/3/26

data = {
    "Customer" : ["Aman","priya","rahul","sneha","psp"],
    "Age" : [25,np.nan,30,np.nan,10],
    "City" : ["Delhi",None,"Delhi","Bangalore","Bangalore"] # np.nan or none both will be replaced
}

df = pd.DataFrame(data)
print(df)
```

```python
print(df.isnull())
print(df.isnull().sum())
```

```python
# print(df.dropna(axis = 0, how = "any")) # if there is any NaN or None values in row then it will remove whole row
print(df.dropna(axis = 0, how = "all")) # if there are all NaN or None values in row then it will remove whole row
```

```python
# print(df.dropna(axis = 1, how = "any")) # if there is any NaN or None values in column then it will remove whole column
print(df.dropna(axis = 1, how = "all")) # if there are all NaN or None values in column then it will remove whole column
```

```python
print(df["Age"].fillna(df["Age"].mean())) # it will fill all NaN or none values with the mean of all age values
print(df["Age"].fillna(df["Age"].median())) # it will fill all NaN or none values with the median of all age values
```

```python
# 28/3/26

df["City"] = df["City"].fillna(df["City"].mode()[0]) # it replaces the categorical data nan/none values with the particular index of mode values 
df
```

```python
df["City"] = df["City"].fillna("Unknown")
print(df)
```

```python
df = df.fillna({
    "Age": df["Age"].mean(),
    "City": "Unknown"
})
print(df)
```

```python
df.drop_duplicates() # to remove the duplicate rows from the dataset
```

```python
sales = pd.Series([100,120,130,110,115,500])

Q1 = sales.quantile(0.25)
Q3 = sales.quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = sales[(sales < lower) | (sales > upper)]

outliers
```

```python
mean = sales.mean()
std = sales.std()

z_scores = (sales - mean) / std

z_scores
```

```python
data = {
    "Date" : ["2024-01-01","2024-2-15","2024-3-10"] # it is in the form of string data and not date format so we convert it to date format with pd.to_datetime function
    # "Date1" : [2024-1-1,2024-2-15,2024-3-10] 
}

df = pd.DataFrame(data)
# df.info() # it will give output as string for Date and int for Date1
df["Date"] = pd.to_datetime(df["Date"])
df
```

```python
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Day"] = df["Date"].dt.day

df
```

```python
data = {
    "Start_Date" : ["2023-1-1","2023-6-1"],
    "End_Date" : ["2024-1-1","2024-6-1"]
}

df = pd.DataFrame(data)

df["Start_Date"] = pd.to_datetime(df["Start_Date"])
df["End_Date"] = pd.to_datetime(df["End_Date"])

df["Tenure_Days"] = (df["End_Date"] - df["Start_Date"]).dt.days
df
```
