```python
# Update your assign_clusters code to work with Manhattan distance as well. Allow the user to choose between 'euclidean' and 'manhattan' as a parameter.

import math

def euclidean(p1,p2):
    return math.sqrt((p1[0]-p2[0])**2+(p1[1]-p2[1])**2)

def manhattan(p1,p2):
    return abs(p1[0]-p2[0])+abs(p1[1]-p2[1])

def assign_clusters(points,centroids,metric="euclidean"):

    assignments=[]

    for point in points:

        distances=[]

        for centroid in centroids:

            if metric=="euclidean":
                d=euclidean(point,centroid)

            elif metric=="manhattan":
                d=manhattan(point,centroid)

            else:
                raise ValueError("Choose euclidean or manhattan")

            distances.append(d)

        assignments.append(distances.index(min(distances)))

    return assignments


points=[(2,3),(5,8),(1,2),(6,9),(7,7)]

centroids=[(2,3),(6,9)]

print("Euclidean")

print(assign_clusters(points,centroids,"euclidean"))

print()

print("Manhattan")

print(assign_clusters(points,centroids,"manhattan"))
```
