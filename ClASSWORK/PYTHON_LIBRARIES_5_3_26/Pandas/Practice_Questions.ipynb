```python
# Questions from Gmail file

# 1) Create a DataFrame with columns: Name, Age, City, Salary

data = {
    "Name" : ["A","B","C","D"],
    "Age" : [20,21,22,23],
    "City" : ["Ahmedabad","Mumbai","Pune","Hyderabad"],
    "Salary" : [12000,13000,14000,15000]
}

df = pd.DataFrame(data)

# Tasks:
# • Display only the Age column
df["Age"]

# • Display Name and Salary together
df[["Name","Salary"]]
```

```python
# 2) Using the same dataset:
# • Select the 3rd row using iloc
df.iloc[2]

# • Select the Salary of the 2nd row using loc
df.loc[1,"Salary"]
```

```python
# 3) Filter the dataset:
# • Show rows where Salary > 13000
df[df["Salary"] > 13000]

# • Show rows where City is "Pune"
df[df["City"] == "Pune"]
```

```python
# 4) Apply multiple conditions:
# • Salary > 40000 AND City = "Mumbai"
df[(df["Salary"] > 13000) & (df["City"] == "Hyderabad")]
```

```python
# 5) Sort the dataset:
# • Sort by Salary in ascending order
df.sort_values("Salary")
# • Sort by Salary in descending order
sorted = df.sort_values("Salary",ascending=False)
sorted
```

```python
# 6) After sorting:
# • Reset the index
sorted.reset_index()
# • Remove the old index
new_sorted = sorted.reset_index(drop=True) # to drop old indexes and save the new ones we need to save it in another variable
new_sorted
```

```python
print(sorted.loc[1])
print(new_sorted.loc[1])
```

```python
# 7) Create a dataset with missing values in Age.

data = {
    "Name" : ["A","B","C","D"],
    "Age" : [20,np.nan,22,None],
    "City" : ["Ahmedabad",None,"Pune","Hyderabad"],
    "Salary" : [12000,13000,14000,15000]
}

df = pd.DataFrame(data)

# Tasks:
# • Detect missing values using isnull()
print(df.isnull())
# • Count missing values per column
print(df.isnull().sum())
```

```python
# 8) Using the same dataset:
# • Drop rows with missing values
# print(df.dropna(axis = 0, how = "any"))
# • Drop columns with missing values
# print(df.dropna(axis = 1, how = "any"))
df
```

```python
# 9) Fill missing values:
# • Replace missing Age with mean
print(df["Age"].fillna(df["Age"].mean()))
# • Replace missing City with "Unknown"
df["City"] = df["City"].fillna("Unknown")
df
```
