```python
import matplotlib.pyplot as plt
```

```python
# SCATTER PLOT

hrs = [1,2,3,4,5,6]
marks = [40,50,60,70,80,90]

plt.scatter(hrs,marks,color="green")

plt.title("Study hrs vs marks")
plt.xlabel("Hours Studied")
plt.ylabel("Marks Obtained")

plt.show()
```

```python
# BAR GRAPH/PLOT

products = ["A","B","C","D"]
sales = [450,500,650,700,]

plt.bar(products,sales,color = "green",width = 0.5)

plt.title("Sales by products")
plt.xlabel("Products")
plt.ylabel("Sales")

plt.show()
```

```python
# HISTOGRAM

marks = [45,50,60,65,70,80,85,90,95,50,60,70]

plt.hist(marks,bins = 1, color = "Purple", edgecolor = "green") # bins value is the number of divisions in which the whole data will be equally divided

plt.title("Marks distribution")
plt.xlabel("Marks range")
plt.ylabel("Frequency")

plt.show()
```

```python
x = [1,2,3,4,5]
y = [10,20,15,25,30]

plt.plot(x,y,color="blue")
plt.scatter(x,y)
plt.bar(x,y,color="red")

plt.title("Using Figure and Axes")
plt.xlabel("X Values")
plt.ylabel("Y Values")

plt.show()
```

```python
x = [1,2,3,4]
y1 = [10,20,15,25]
y2 = [20,30,35,45]

plt.plot(x,y1,label="X & Y1 Data")
plt.plot(x,y2,label="X & Y2 Data")

plt.title("Multiple Line Chart")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.legend() # it shows the info of graph

plt.show()
```
