# SESSION 5 — Handling Missing Values in Pandas

## Why Handling Missing Values is Important

In real-world datasets, it is very common to encounter missing values.

Examples:

| Customer | Age | City |
|----------|-----|------|
| Aman | 25 | Delhi |
| Priya | NaN | Mumbai |
| Rahul | 30 | Delhi |

Missing values can occur due to:

• Data entry errors  
• Incomplete forms  
• System failures  
• Data collection issues  

If missing values are not handled properly, they can cause:

• Incorrect analysis  
• Errors in machine learning models  
• Biased results  

Therefore data cleaning is a critical step in data analysis.

---

# Identifying Missing Values

Pandas represents missing values using **NaN**.

Function used:

isnull()

## Syntax

df.isnull()

## Purpose

Returns a boolean DataFrame where:

True → value is missing  
False → value is present

```python
import pandas as pd
import numpy as np

data = {
    "Customer": ["Aman", "Priya", "Rahul", "Sneha"],
    "Age": [25, np.nan, 30, np.nan],
    "City": ["Delhi", "Mumbai", "Delhi", "Bangalore"]
}

df = pd.DataFrame(data)

df
```

```python
df.isnull()
```

# Counting Missing Values

To count missing values per column we use:

df.isnull().sum()

```python
df.isnull().sum()
```

# Handling Missing Values in Categorical Columns

Pandas supports multiple data types in a DataFrame, such as:

• Numerical → Age, Salary  
• Categorical → City, Gender, Product  
• Boolean → True/False  
• Datetime → Dates  

Missing values (NaN) can exist in **any type of column**, not just numerical.

---

# Important Clarification

• dropna() works on all columns  
• fillna() works on all columns  

BUT

The way we **fill missing values depends on the data type**

---

# Handling Numerical vs Categorical Data

Numerical Columns:

• Replace with mean  
• Replace with median  

Categorical Columns:

• Replace with mode (most frequent value)  
• Replace with constant like "Unknown"

---

# Why Mean Does Not Work for Categorical Data

Example:

City column:

Delhi, Mumbai, NaN

We cannot calculate mean of "Delhi" and "Mumbai"

So we use:

• Mode (most frequent value)  
• Constant value like "Unknown"

```python
import pandas as pd
import numpy as np

data = {
    "Customer": ["Aman", "Priya", "Rahul", "Sneha"],
    "Age": [25, np.nan, 30, np.nan],          # Numerical
    "City": ["Delhi", np.nan, "Delhi", "Mumbai"]  # Categorical
}

df = pd.DataFrame(data)

print("Original DataFrame:\n")
print(df)
```

# 1. Filling Numerical Column (Age)

Use mean or median


```python
df["Age"] = df["Age"].fillna(df["Age"].mean())

print(df)
```

# 2. Filling Categorical Column Using Mode

Mode = most frequent value

```python
df["City"] = df["City"].fillna(df["City"].mode())

print(df)
```

# 3. Filling Categorical Column Using Constant

Instead of mode, we can use a constant value.

Example:

"Unknown"

```python
df = pd.DataFrame(data)

df["City"] = df["City"].fillna("Unknown")

print(df)
```

# 4. Using fillna() for Multiple Columns

We can handle both numerical and categorical columns together.

Example:

• Age → replace with mean  
• City → replace with "Unknown"

```python
df = pd.DataFrame(data)

df = df.fillna({
    "Age": df["Age"].mean(),
    "City": "Unknown"
})

print(df)
```

# What are Outliers?

Outliers are extreme values that differ significantly from other observations.

Example:

[10, 12, 11, 13, 500]

Value **500** is an outlier.

Outliers can occur due to:

• measurement errors  
• data entry errors  
• unusual events

# Detecting Outliers using IQR Method

IQR = Interquartile Range

Formula:

IQR = Q3 − Q1

Lower Bound = Q1 − 1.5 × IQR  
Upper Bound = Q3 + 1.5 × IQR

```python
sales = pd.Series([100,120,130,110,115,500])

Q1 = sales.quantile(0.25)
Q3 = sales.quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = sales[(sales < lower) | (sales > upper)]

outliers
```

# Z-Score Concept

Z-score measures how far a value is from the mean.

Formula:

Z = (X − Mean) / Standard Deviation

Values with |Z| > 3 are often considered outliers.

```python
mean = sales.mean()
std = sales.std()

z_scores = (sales - mean) / std

z_scores
```

# SESSION 7 — Column Operations & Data Transformations

In data analysis we often need to:

• Create new columns  
• Modify existing columns  
• Apply functions to columns

Pandas provides several methods to perform transformations.

# Adding New Columns

A new column can be created using assignment.

Example:

Revenue = Price × Quantity

```python
import pandas as pd

data = {
    "Product": ["Pen", "Notebook", "Marker"],
    "Price": [10, 50, 20],
    "Qty": [100, 40, 60]
}

df = pd.DataFrame(data)

df
```

```python
df["Revenue"] = df["Price"] * df["Qty"]

df
```

# Using apply()

apply() allows applying a function to columns.

## Syntax

df["column"].apply(function)

Commonly used for custom transformations.

```python
df["Price_with_tax"] = df["Price"].apply(lambda x: x * 1.18)

df
```

# Using lambda Functions

Lambda functions are small anonymous functions.

Syntax:

lambda argument : expression

# Using applymap()

applymap() applies a function to **every element of the DataFrame**.

## Syntax

df.applymap(function)

Note:

Used only for DataFrames (not Series).

```python
df_numeric = df[["Price","Qty"]]

# df_numeric.applymap(lambda x: x * 2)
df_numeric.map(lambda x: x * 2)
```
