# Session 1: Introduction to Python Libraries

## 1. Why Do We Need Libraries?

When solving real-world computational problems, we often require:

- Mathematical operations on large datasets
- Statistical computations
- Data manipulation and cleaning
- Visualization of results
- Efficient memory usage

---

## 2. What is a Python Library?

A Python library:

- Contains modules and functions
- Is written by experts and optimized for performance
- Can be imported and used in our programs
- Prevents code repetition

# Difference Between Built-in Functions, Modules, and Libraries

In Python, functionality is organized at different levels.  
To use Python efficiently, it is important to clearly understand the structural difference between:

- Built-in Functions  
- Modules  
- Libraries  

---

## 1. Built-in Functions

Built-in functions are:

- Predefined functions available by default
- Immediately usable without importing anything
- Part of Python’s core

They perform fundamental operations such as:
- Printing output
- Type conversion
- Length calculation
- Mathematical operations

### Examples:

- `print()`
- `len()`
- `type()`
- `sum()`
- `max()`

### Key Characteristics:

- No import statement required
- Directly accessible
- Limited to fundamental operations


````python
# Examples of built-in functions

numbers = [10, 20, 30]

print("Length:", len(numbers))
print("Type:", type(numbers))
print("Sum:", sum(numbers))
````

## 2. Modules

A module is:

- A single Python file containing related functions and variables
- Designed to organize code logically
- Requires import before use

Modules extend Python’s capabilities beyond built-in functions.

### Example:

The `math` module provides advanced mathematical operations.

### Key Characteristics:

- Must be imported
- Contains related functions grouped together
- Improves modularity and code organization


````python
# Example of a module

import math

print("Square root:", math.sqrt(25))
print("Factorial:", math.factorial(5))
````

## 3. Libraries

A library is:

- A collection of multiple modules
- Designed to solve broader and more complex problems
- Built for specialized domains such as data analysis, visualization, or machine learning

Libraries are structured collections of modules working together.

### Examples:

- NumPy → Numerical computation
- Pandas → Data manipulation
- Matplotlib → Data visualization

### Key Characteristics:

- Contain multiple modules
- Designed for domain-specific tasks
- Provide high-level functionality
- Must be imported before use


````python
# Example of using a library

import numpy as np

array = np.array([1, 2, 3, 4])
print("Array multiplied by 2:", array * 2)
````

# Session 2: Core Data Structures

Before learning advanced libraries, we must understand Python data structures.

## 1. List

A list is:
- Ordered
- Mutable
- Allows duplicate values
- Indexed

Example scenario:
Store marks of students.


````python
# Creating a list
marks = [78, 85, 90, 85, 88]

# Accessing elements
print("First element:", marks[0])

# Modifying element
marks[1] = 95

# Appending new value
marks.append(92)

print("Updated list:", marks)
````

### Step-by-step explanation:

1. Square brackets define a list.
2. Indexing starts from 0.
3. Lists allow modification.
4. Duplicate values are allowed.


## 2. Tuple

A tuple is:
- Ordered
- Immutable
- Allows duplicates
- Indexed

### Used when data should remain constant.
### Example: Geographic coordinates.


````python
coordinates = (28.7041, 77.1025)

print("Latitude:", coordinates[0])

# Uncommenting the next line will cause an error
# coordinates[0] = 30
````

### Key Observation:
- Tuples do not allow modification after creation.
- This ensures data integrity.


## 3. Set

A set is:
- Unordered
- Mutable
- Does not allow duplicates
- Not indexed

Used when we need unique elements.


````python
numbers = {1, 2, 3, 3, 4, 5, 5}

print("Set output:", numbers)

# Adding an element
numbers.add(10)

print("Updated set:", numbers)
````

### Key Observation:
- Duplicate values are automatically removed.
- Sets are useful for membership testing.

# Session 3: Introduction to NumPy

## 1. Limitations of Python Lists

Consider a task:
Add 5 to every element in a list.

Using traditional Python:


````python
numbers = [1, 2, 3, 4]

result = []
for i in numbers:
    result.append(i + 5)

print(result)
````

For small datasets, this approach is acceptable.

However, for millions of elements:
- Loop execution becomes slow
- Memory usage increases
- Performance decreases

To solve this, we use NumPy.


## 2. What is NumPy?

NumPy stands for Numerical Python.

It is used for:
- Numerical computation
- Matrix operations
- Scientific computing
- High-performance array processing


````python
import numpy as np

arr = np.array([1, 2, 3, 4])

print("Original array:", arr)
print("After adding 5:", arr + 5)
````

### Explanation:

1. NumPy arrays support vectorized operations.
2. No explicit loop is required.
3. Operations are faster than lists.
4. NumPy uses optimized C-based backend.


````python
matrix = np.array([[1, 2, 3],
                   [4, 5, 6]])

print("Matrix:")
print(matrix)

print("Matrix multiplied by 2:")
print(matrix * 2)
````

# Is Dictionary a Prerequisite for Learning Pandas?

## Short Answer

A dictionary is not strictly mandatory before learning Pandas.

However, understanding dictionaries makes learning Pandas significantly easier.

---

# 1. Why Dictionary Knowledge Helps in Pandas

Pandas internally represents tabular data in a structure similar to:

- Keys → Column Names
- Values → Column Data

This is conceptually very similar to a Python dictionary.

Dictionary structure:

{
    "column_name": [values]
}

Pandas DataFrame can be created directly from a dictionary.

---

# 2. Quick Revision: Dictionary Structure

A dictionary in Python:

- Stores data in key-value pairs
- Keys must be unique
- Values can be lists, numbers, strings, etc.
- Defined using curly braces {}

Example:


````python
student_data = {
    "Name": ["Aman", "Riya", "Kunal"],
    "Age": [21, 20, 22],
    "Marks": [88, 92, 79]
}

print(student_data)
````

Here:

- "Name", "Age", "Marks" are keys
- Each key maps to a list of values

This structure is directly compatible with Pandas.

---

# 3. Creating a DataFrame from Dictionary


````python
import pandas as pd

student_data = {
    "Name": ["Aman", "Riya", "Kunal"],
    "Age": [21, 20, 22],
    "Marks": [88, 92, 79]
}

df = pd.DataFrame(student_data)

print(df)
````

Notice:

Each dictionary key becomes a column.
Each list becomes the column values.

Therefore:

Dictionary understanding makes DataFrame creation intuitive.

---

# 4. Is Dictionary Compulsory Before Pandas?

Not strictly.

Because:

- Pandas can read CSV files directly.
- Pandas can read Excel files.
- Pandas can load JSON data.

However:

To understand:
- Column selection
- DataFrame structure
- Internal representation

Dictionary knowledge is strongly recommended.

---

# 5. Conceptual Mapping

Dictionary Concept            →        Pandas Equivalent

Key                          →        Column Name  
Value (List)                 →        Column Data  
Key-Value Collection         →        Entire DataFrame  

---

# 6. Final Conclusion

Dictionary is not a strict prerequisite.

But:

Understanding dictionary makes:

- DataFrame creation easier
- Column operations clearer
- Conceptual understanding stronger

# Session 4: Introduction to Pandas

## 1. Need for Structured Data Handling

Real-world datasets contain:
- Rows and columns
- Mixed data types
- Missing values
- Large volume

Python lists and NumPy arrays are not sufficient for structured tabular data.

Pandas solves this problem.


````python
import pandas as pd

data = {
    "Name": ["Aman", "Riya", "Kunal"],
    "Age": [21, 20, 22],
    "Marks": [88, 92, 79]
}

df = pd.DataFrame(data)

print(df)
````

## 2. Selecting Columns


````python
print(df["Marks"])
````

## 3. Filtering Data


````python
high_marks = df[df["Marks"] > 85]
print(high_marks)
````

## 4. Basic Statistics


````python
print("Average Marks:", df["Marks"].mean())
print("Maximum Marks:", df["Marks"].max())
````

# Session 5: Data Visualization using Matplotlib

Numerical output alone is difficult to interpret.
Visualization helps in:

- Identifying trends
- Comparing values
- Presenting insights clearly


````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [10, 20, 25, 30]

plt.plot(x, y)
plt.title("Line Plot Example")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
````

````python
students = ["Aman", "Riya", "Kunal"]
marks = [88, 92, 79]

plt.bar(students, marks)
plt.title("Marks Distribution")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
````

# Complete Data Workflow Example

Step 1: Create Data  
Step 2: Convert to DataFrame  
Step 3: Perform Analysis  
Step 4: Visualize Results


````python
import pandas as pd
import matplotlib.pyplot as plt

# Step 1: Create dataset
data = {
    "Student": ["Aman", "Riya", "Kunal", "Sahil"],
    "Marks": [78, 85, 90, 88]
}

# Step 2: Convert to DataFrame
df = pd.DataFrame(data)

# Step 3: Calculate average
average = df["Marks"].mean()
print("Average Marks:", average)

# Step 4: Visualize
plt.bar(df["Student"], df["Marks"])
plt.title("Student Performance")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.show()
````
