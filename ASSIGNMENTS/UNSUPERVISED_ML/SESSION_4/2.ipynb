```python
# Plot the 2D projection of the Iris dataset after PCA transformation, using a different color for each species, similar to how Spotify visualizes genre clusters.<br><br><em><strong>Hint:</strong> Use matplotlib's scatter() and color by target label.</em>

import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.decomposition import PCA

# Load dataset
iris = load_iris()
X = iris.data
y = iris.target

# PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

# Colors
colors = ['red', 'green', 'blue']

plt.figure(figsize=(8,6))

for i, color in enumerate(colors):
    plt.scatter(
        X_pca[y == i, 0],
        X_pca[y == i, 1],
        color=color,
        label=iris.target_names[i]
    )

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA Projection of Iris Dataset")
plt.legend()
plt.grid(True)

plt.show()
```
