```python
# Use the Elbow Method to determine the optimal number of clusters (k) for a dataset of your choice (e.g., favorite songs by tempo and danceability, or Zomato restaurant ratings and price). Plot the within-cluster sum of squares for k=1 to k=6 and identify the 'elbow' point.<br><br><em><strong>Hint:</strong> Use sklearn's KMeans and matplotlib for plotting.</em>

import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

songs = [
    [120,0.80],
    [118,0.75],
    [90,0.30],
    [95,0.35],
    [130,0.90],
    [135,0.88],
    [80,0.20],
    [85,0.25],
    [110,0.60],
    [125,0.82]
]

wcss=[]

for k in range(1,7):

    kmeans=KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(songs)

    wcss.append(kmeans.inertia_)

plt.figure(figsize=(7,5))

plt.plot(range(1,7),wcss,marker='o')

plt.xlabel("Number of Clusters (k)")
plt.ylabel("WCSS")

plt.title("Elbow Method")

plt.grid(True)

plt.show()

print("WCSS Values")

for i,value in enumerate(wcss,start=1):
    print(f"k={i}: {value:.2f}")
```
