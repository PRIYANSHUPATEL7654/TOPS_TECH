```python
from dotenv import load_dotenv
import os
load_dotenv()
```

```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict # data gets in python dictionary with this 
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-3.5.flash",
    temperature=0
    )
```

```python
class State(TypedDict): # this is state class where we store all our variables such as name and this class will be accessed by nodes in this file
    question:str
    answer:str 

# inheritance : child class can access the function of parent class
```

```python
def question_node(state:State):
    print("Question node running...")
    return{
        "question":state["question"]
    }
```

```python
def llm_node(state:State):
    print("LLM node running...")
    response = llm.invoke(state["question"])
    return{
        "answer":response.content
    }
```

```python
def response_node(state:State):
    print("Response node running...")
    print("Question : ",state["question"])
    print("Answer : ",state["answer"])
    return{
        "question":state["question"],
        "answer":state["answer"]
    }
```

```python
graph = StateGraph(State)
```

```python
graph.add_node("question",question_node)
graph.add_node("llm",llm_node)
graph.add_node("response",response_node)
```

```python
graph.add_edge(START,"question")
graph.add_edge("question",llm)
graph.add_edge("llm","response")
graph.add_edge("response",END)
```

```python
app = graph.compile()
```

```python
result = app.invoke({
    "question" : "What is Python?",
    "answer":""
})
print(result["answer"])
```
