```python
import matplotlib.pyplot as plt
```

```python
hrs = [1,2,3,4,5,6]
marks = [40,50,55,65,70,80]

plt.scatter(hrs,marks,color = "#000000")

plt.title("Study Hrs vs Marks")
plt.xlabel("Hrs")
plt.ylabel("Marks")
plt.show()
```

```python
age = [20,22,24,26,28]
income = [20000,25000,30000,35000,40000]

plt.scatter(age,income,color = "#000000")

plt.title("Age vs Income")
plt.xlabel("Age")
plt.ylabel("Income")
plt.show()
```

```python
products = ['a','b','c','d']
sales = [100,150,80,200]

plt.bar(products,sales,color = "blue")
plt.title("Products vs Sales")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.show()
```
