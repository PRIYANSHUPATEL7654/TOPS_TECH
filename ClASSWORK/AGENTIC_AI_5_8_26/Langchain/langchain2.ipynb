```python
from dotenv import load_dotenv # to import the gemini api key
import os

#prompt template - langchain - llm
# dynamic python - graph
# reuse and dynamic
# Prompt Template : Using prompt template we can fix the number of questions/answers and also the level of questions/answers about the topic given by the user.
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate
```

```python
load_dotenv()
```

```python
llm=ChatGoogleGenerativeAI(

    model="gemini-3.5-flash",
    temperature=0.7
)
```

```python
tech = PromptTemplate(
    input_variables=["tech"], # this word will be used as dynamic in the prompt template and this is the word the model will take as input from the user.
    template="""
You are an experienced technical interviewer.
Generate 5 interview questions for {tech}.
Requirements:
1. Generate 3 basic questions.
2. Generate 2 advanced questions.
3. Keep the questions clear and practical.
"""
)
```

```python
technology = input("Enter your technology : ")
```

```python
prompt = tech.format(tech=technology)
```

```python
response = llm.invoke(prompt)
# llm request - llm response
```

```python
print(response.content)
```
