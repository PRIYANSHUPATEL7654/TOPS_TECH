```python
import numpy as np 
from tensorflow.keras.preprocessing.text import Tokenizer 
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.models import Sequential 
from tensorflow.keras.layers import Embedding, LSTM, Dense
```

```python
with open("mydata.txt","r") as file:
    text = file.read()
    text = text.lower()
    print(text)
```

```python
tokenizer = Tokenizer()
tokenizer.fit_on_texts([text]) # it will give numbers to all texts
print(tokenizer.word_index)
```

```python
sequence = tokenizer.texts_to_sequences([text])[0] # the tokenizer will give numbers to all words but it will start with one and at the end we have to store it in a list and it starts with 0 so here we add 0 first to the sequence
print(sequence)
```

```python
X = []
y = []

for i in range(2,len(sequence)):
    X.append(sequence[i-2:i])
    y.append(sequence[i])

# https://chatgpt.com/share/6a6b7e68-6180-83ee-a876-9ea34e25f990

X = np.array(X)
y = np.array(y)
```

```python
total_words = len(tokenizer.word_index) + 1
y = to_categorical(y,num_classes=total_words)
```

```python
model = Sequential()
model.add(Embedding(total_words, 10, input_length = 2))
model.add(LSTM(20))
model.add(Dense(total_words, activation = "softmax"))
```

```python
model.compile(
    optimizer="adam",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
model.fit(X,y,epochs=500,verbose=1)
```

```python
seed_text = "Machine Learning"

seed_text = seed_text.lower()

encoded = tokenizer.texts_to_sequences([seed_text])[0]

encoded = np.array(encoded).reshape(1, 2)

prediction = model.predict(encoded, verbose=0)

predicted_word = tokenizer.index_word[np.argmax(prediction)]

print("Next Word :", predicted_word)
```
