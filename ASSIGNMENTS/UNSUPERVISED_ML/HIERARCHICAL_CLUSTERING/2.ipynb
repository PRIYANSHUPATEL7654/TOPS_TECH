```python
# Using Python and sklearn, perform agglomerative hierarchical clustering on a dataset of 8 songs (with features like tempo and energy), and plot the resulting dendrogram using scipy's dendrogram function.

import numpy as np
import matplotlib.pyplot as plt

from scipy.cluster.hierarchy import linkage,dendrogram

songs=np.array([

[120,0.80],

[118,0.75],

[90,0.30],

[95,0.35],

[130,0.90],

[135,0.88],

[80,0.20],

[85,0.25]

])

Z=linkage(
songs,
method='ward'
)

plt.figure(figsize=(10,6))

dendrogram(
Z,
labels=['S1','S2','S3','S4','S5','S6','S7','S8']
)

plt.title("Agglomerative Hierarchical Clustering")

plt.xlabel("Songs")

plt.ylabel("Distance")

plt.grid(True)

plt.show()
```
