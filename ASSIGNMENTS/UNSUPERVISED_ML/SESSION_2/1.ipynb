```python
import matplotlib.pyplot as plt

points = [(2,3),(5,8),(1,2),(6,9),(7,7)]

centroids = [(2,3),(6,9)]

x = [p[0] for p in points]
y = [p[1] for p in points]

plt.scatter(x,y,color='blue',label='Food Delivery Locations')

plt.scatter(
    [centroids[0][0],centroids[1][0]],
    [centroids[0][1],centroids[1][1]],
    color='red',
    marker='X',
    s=200,
    label='Initial Centroids'
)

for i,p in enumerate(points):
    plt.text(p[0]+0.1,p[1]+0.1,f"P{i+1}")

plt.xlabel("X Coordinate")
plt.ylabel("Y Coordinate")
plt.title("Food Delivery Locations")
plt.legend()
plt.grid(True)
plt.show()
```
