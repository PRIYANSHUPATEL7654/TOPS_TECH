```python
import mysql.connector
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.tools import tool
from langchain.agents import create_agent
```

```python
load_dotenv()
```

```python
db = mysql.connector.connect(
    host="localhost",
    user="root",
    password="PSP76542003",
    database="advanced_sql",
    port=3306
)
```

```python
print(db.is_connected())
```

```python
llm = ChatGoogleGenerativeAI(
    model = "gemini-3.5-flash",
    temperature = 0
)
```

```python
@tool
def get_student_data():
    """Get all student data from MySQL database."""

    cursor = db.cursor()
    cursor.execute("SELECT * FROM STUDENT_DETAILS")

    data = cursor.fetchall()

    return data
```

```python
agent = create_agent(
    model=llm,
    tools=[get_student_data],
    system_prompt="""
    You are a helpful student assistant.
    
    Use the psp database tool whenever the user asks about students.

    Do not make up student information.

    Use the database data to answer.
    """
)
```

```python
question = input("Ask your question : ")
```

```python
result = agent.invoke({
    "messages" : [
        ("user",question)
    ]
})
```
