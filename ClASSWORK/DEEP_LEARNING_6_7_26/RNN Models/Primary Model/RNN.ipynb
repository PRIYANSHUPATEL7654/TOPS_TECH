```python
import numpy as np 
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,SimpleRNN, Dense
```

```python
X = np.array([
    [1,2], # I Love
    [3,4], # Virat Plays
    [5,6], # MS Dhoni
    [7,8]  # We Watch
])

Y = np.array([
    9,  # cricket
    9,  # cricket
    10, # captain
    11  # football
])
```

```python
word = {
    0:'I',
    1:'Love',
    2:'Virat',
    3:'Plays',
    4:'MS',
    5:'Dhoni',
    6:'We',
    7:'Watch',
    8:'cricket',
    9:'cricket',
    10:'captain',
    11:'football'
}
```

```python
model = Sequential()

model.add(Embedding(input_dim=12, output_dim=8)) # https://chatgpt.com/c/6a60ae31-2f08-83ee-9ab7-514280d5d201

model.add(SimpleRNN(16)) # 16 Hidden units (neurons)

model.add(Dense(12, activation="softmax"))
```

```python
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
model.fit(X,Y,epochs=300)
print("Training Completed")
```

```python
test = np.array([[3,4]])
prediction = model.predict(test)
index = np.argmax(prediction)
print(index)
```

```python
print("Next word : ",word[index])
```
