```python
# Generate a heatmap of the correlation matrix for the numerical features in the dataset using seaborn, and interpret which two features are most strongly correlated.

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("Mall_Customers.csv")

# Select only numerical columns
numeric_df = df.select_dtypes(include=['int64', 'float64'])

# Correlation Matrix
correlation = numeric_df.corr()

# Plot Heatmap
plt.figure(figsize=(8,6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()
```
