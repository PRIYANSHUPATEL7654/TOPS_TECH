```python
# Repeat the dendrogram plotting for the same data using 'single' and 'complete' linkage methods, and compare how the cluster shapes and heights differ from 'ward' linkage.<br><br><em><strong>Hint:</strong> Place all three dendrograms side by side for visual comparison.</em>

from sklearn.datasets import load_iris
from scipy.cluster.hierarchy import linkage, dendrogram
import matplotlib.pyplot as plt

iris = load_iris()
X = iris.data

methods = ['single', 'complete', 'ward']

plt.figure(figsize=(18,5))

for i, method in enumerate(methods):

    plt.subplot(1,3,i+1)

    Z = linkage(X, method=method)

    dendrogram(
        Z,
        no_labels=True
    )

    plt.title(method.capitalize() + " Linkage")

plt.tight_layout()
plt.show()
```
