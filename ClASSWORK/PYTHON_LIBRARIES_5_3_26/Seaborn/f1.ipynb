```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
df = sns.load_dataset("tips")
df.head(20)
```

```python
df.info()
```

```python
sns.histplot(data = df,x = "total_bill", bins = 10,kde = True)
sns.histplot(data = df,x = "total_bill", bins = 40)
# plt.show() 
```

```python
sns.kdeplot(data = df,x = "total_bill")
# plt.show()
```

```python
sns.boxplot(x="day",y="total_bill",data=df)
plt.show()
# the box represents maximum no of values ,line shows other values and circles show outliers in the dataset
```

```python
sns.violinplot(data=df,x="sex", y="total_bill")
plt.show()
```
