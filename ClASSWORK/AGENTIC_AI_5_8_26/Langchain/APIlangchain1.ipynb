```python
import requests
# the requests library helps in connecting python with http
from dotenv import load_dotenv # to connect with env file
from langchain_google_genai import ChatGoogleGenerativeAI # to connect with llm
```

```python
# langchain considers everything as a tool so to make our function tool and be accessible by langchain with use @annotation in front of function to make it a tool for langchain
from langchain.tools import tool
# to use langchain as an agent 
from langchain.agents import create_agent

# When we create an agent we will first pass its version and its temperature
# After creating an agent firstly we pass this llm model and will pass this info stored in it
# The 2nd thing we will pass is tools (@tool which we wrote above function, will pass that function in this tool)
# The 3rd thing we will pass is our prompt
#The tool will store data in list as there might be more than 1 tools working together
```

```python
load_dotenv()
```

```python
llm = ChatGoogleGenerativeAI(
    model = "gemini-3.5-flash",
    temperature = 0.7
)
```

```python
API1 = "https://6a7d92cbf8b2ed99ca4e8036.mockapi.io/api/v1/electronicsapi"
API2 = "https://681444af225ff1af16285b31.mockapi.io/furnitureapi"
```

```python
@tool # annotation
def mydataelectronics():
    """give me all electronics product from this api""" # this is called instruction docstring, agent will first see this when function is called

    response = requests.get(API1)
    return response.json()
```

```python
@tool
def mydatafurniture():
    """give me all furniture product from this api""" 

    response = requests.get(API2)
    return response.json()
```

```python
result1 = mydataelectronics.invoke({})
result1
```

```python
result2 = mydatafurniture.invoke({}) # The response we will get will be in json, so to store it appropriately we pass dictionary here
result2
```

```python
agent = create_agent(
    model = llm,
    tools = [mydataelectronics,mydatafurniture], # here are two different functions in which the data is coming from two different apis so i can also write only one api and its function here if i want only one.
    system_prompt="""You are a helpful electronics and furniture assistant
    
    Use the electronics tool whenever the user asks about available electronics products.

    Use the furniture tool whenever the user asks about available furniture products.

    Do not make up product information.
    Use the tool data to answer"""
    # here the prompt should be such that it covers all the tools we have passed so that it can respond appropriately
)
```

```python
question = input("Ask me anything...")

result = agent.invoke({

    "messages":[
        ("user",question) # this is a tuple
    ]
})
```

```python
print(result["messages"][-1].content[0]["text"])
# -1 means the last message, here from passing the message from langchain to llm and back, there are series of messages but we want only the output and the output is always at last so that's the reason of passing -1 here.
# content means extracting the content of that last message, and there can be multiple messages and to get it we write indexes of it as per our need, here 0 means the first content block of that last message and text written here is the key and its values will be passed as output.

# https://chatgpt.com/share/6a7ef7bc-9344-83e8-9aed-937bfa2daf20
```
