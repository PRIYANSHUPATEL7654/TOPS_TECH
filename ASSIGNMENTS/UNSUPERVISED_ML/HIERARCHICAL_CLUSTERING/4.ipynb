```python
# Write a function that, given two points (x1, y1) and (x2, y2), returns both their Euclidean and Manhattan distances. Test it on three pairs of coordinates from your dataset.

import math

def calculate_distances(point1, point2):
    """
    Returns Euclidean and Manhattan distances
    between two points.
    """

    x1, y1 = point1
    x2, y2 = point2

    euclidean = math.sqrt((x2 - x1)**2 + (y2 - y1)**2)

    manhattan = abs(x2 - x1) + abs(y2 - y1)

    return round(euclidean, 2), manhattan


# Food Delivery Dataset
orders = {
    "O1": (10, 2),
    "O2": (12, 3),
    "O3": (15, 5),
    "O4": (18, 8),
    "O5": (20, 9),
    "O6": (25, 12)
}

# Test Cases
pairs = [
    ("O1", "O2"),
    ("O2", "O4"),
    ("O4", "O6")
]

for p1, p2 in pairs:
    e_distance, m_distance = calculate_distances(
        orders[p1],
        orders[p2]
    )

    print(f"{p1} -> {p2}")
    print(f"Euclidean Distance : {e_distance}")
    print(f"Manhattan Distance : {m_distance}")
    print("-" * 35)
```
